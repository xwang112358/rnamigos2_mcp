```python
import json

# Load the JSON file
json_file = "2GDI_Y_TPP_100.json"
with open(json_file, 'r') as f:
    data = json.load(f)
```

```python
# Extract all residues from nodes
all_residues = []
pocket_residues = []

for node in data['nodes']:
    # Parse the ID: format is "2gdi.Y.74" -> we want "Y.74"
    node_id = node['id']
    parts = node_id.split('.')
    chain = parts[1]
    residue_num = parts[2]
    residue = f"{chain}.{residue_num}"
    
    # Add to all residues list
    all_residues.append(residue)
    
    # Add to pocket residues if in_pocket is True
    if node['in_pocket']:
        pocket_residues.append(residue)

# Sort the lists by residue number
all_residues.sort(key=lambda x: int(x.split('.')[1]))
pocket_residues.sort(key=lambda x: int(x.split('.')[1]))

print(f"Total residues: {len(all_residues)}")
print(f"Pocket residues (in_pocket=true): {len(pocket_residues)}")
print()

print("=" * 60)
print("ALL RESIDUES")
print("=" * 60)
print(all_residues)

print("=" * 60)
print("POCKET RESIDUES (in_pocket=true)")
print("=" * 60)
print(pocket_residues)
```

```python
# save pocket residue as npy
import numpy as np

np.save('inputs/residue_list/2gdi.npy', pocket_residues)

```
